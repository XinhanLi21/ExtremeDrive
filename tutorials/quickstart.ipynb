# ExtremeDrive: data, agents and evaluation
Run after `pip install -e .`. No future reference data are passed to the agent. The examples below inspect bundled assets without downloading sensor images.

```python
from pathlib import Path
from extremebench.toolkit.data import Dataset
root = Path.cwd()
if not (root / "assets").exists():
    root = root.parent
ds = Dataset(root / "assets", dataroot=root / "data", split="test")
print(len(ds.tokens()), len(ds.tokens(execution=True)))
```

## Causal planner input
Image paths are resolved under your data root. Download source episodes before opening the images.

```python
token = ds.tokens(execution=True)[0]
inputs = ds.inputs(token)
inputs
```

## Minimal agent
The baseline returns eight future poses, in metres/radians. It does not access the target or evaluation context.

```python
from extremebench.agents import ConstantVelocityAgent
ConstantVelocityAgent().predict(inputs)
```

## Run actual vehicle execution
This executes the three bundled smoke predictions. It does not launch training or inference of a large model.

```python
import subprocess, sys
subprocess.run([sys.executable, "-m", "extremebench.toolkit.cli", "--assets", str(root / "assets"), "evaluate", "--submission", str(root / "examples/smoke_predictions.jsonl"), "--subset", "--out", str(root / "results/tutorial")], check=True)
```
